```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Name': ['Ahmed', 'Ali', 'Nasser', 'Fahad', 'Saleh', 'Sami'],
    'age': [20, 21, 19, 18, 22, 23],
    'Study_Hours': [2, 3, 4, 5, 6, 7],
    'Major': ['Data Science', 'Information Technology', 'Data Science', 'Information Technology', 'Computer Science', 'Information Technology'],
    'grade': [78, 80, 96, 75, 87, 50]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Calculate the average study hours
avg_Hours = df['Study_Hours'].mean()
print(f"Average Study Hours: {avg_Hours}")
```

```python
# true if the student has above average study hours, false otherwise
df['above_avg_Hours'] = df['Study_Hours'] > avg_Hours
print(df)
```
