```python
# import necessary libraries
import pandas as pd
import numpy as np
```

```python
# Create student dataset containing a missing value (NaN)
data = {
    'Name': ['Ahmed', 'Ali','Nasser','Fahad','Saleh','Sami'],
    'age': [20, 21, 19, 18, 22, 23],
    'Study_Hours': [2, 3, 4, 5, 6, 7],
    'Major': ['Data Science', 'Information Technology', 'Data Science', 'Information Technology', 'Computer Science', 'Information Technology'],
    'grade': ['78', '80', '96', np.nan, '87', '50']
}
# Create a DataFrame from the dataset
df = pd.DataFrame(data)
print(df)
print("---------------------------------------------------------------------")
# Filter students by Data Science major
print(df[df.Major == 'Data Science'])
print("---------------------------------------------------------------------")
# Filter grades greater than '90'
print(df[df.grade > '90'])
```

```python
#-------------OPTION 1: Drop rows that contain missing values (NaN)#-------------

#Check and count missing values in the dataframe
print(df.isnull().sum())
# remove rows with missing values
df.dropna(inplace=True)
# print the final cleaned DataFrame
print(df)
```

```python
##-------------OPTION 2: Fill missing values with a specific value#-------------

#Check and count missing values in the dataframe
print(df.isnull().sum())
# Fill missing grade values with 75 instead of dropping the row
df.fillna({'grade': 75}, inplace=True)
#Print the final cleaned DataFrame
print(df)
```
